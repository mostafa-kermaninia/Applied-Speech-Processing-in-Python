```python
import pyttsx3
import os
from IPython.display import Audio, display
import time
import pyttsx3
import os
from IPython.display import Audio, display
import time

# در صورت وجود خطا در نصب، روش آفلاین همچنان کار می‌کند
try:
    from gtts import gTTS
    gtts_ready = True
except ImportError:
    gtts_ready = False
    print("نصب نشد. برای تبدیل آنلاین به اینترنت نیاز دارید، ولی روش آفلاین کامل در دسترس است")

print("تمام کتابخانه‌های اصلی با موفقیت بارگذاری شدند")
```

### 2️⃣ تبدیل متن به گفتار آنلاین (در صورت اتصال به اینترنت بین الملل)


```python
# این سلول فقط در صورتی اجرا می‌شود که اینترنت بین الملل وصل باشد
try:
    text_en = "Hello! Welcome to the speech processing course. Today we will make the computer speak."
    print("📡 در حال اتصال به سرویس گوگل")
    tts = gTTS(text=text_en, lang='en', slow=False)
    tts.save("hello_online.mp3")
    print("خروجی آنلاین (صدای طبیعی گوگل)")
    display(Audio("hello_online.mp3", autoplay=False))
except Exception as e:
    print(f"❌ خطا در اتصال به اینترنت یا سرویس گوگل: {e}")
```

### 3️⃣ تبدیل متن به گفتار آفلاین


```python
def speak_offline(text, rate=200, volume=0.5, voice_index=None):
    engine = pyttsx3.init()
    try:
        engine.setProperty('rate', rate)
        engine.setProperty('volume', volume)
        if voice_index is not None:
            voices = engine.getProperty('voices')
            if 0 <= voice_index < len(voices):
                engine.setProperty('voice', voices[voice_index].id)
        engine.say(text)
        engine.runAndWait()   # حلقه اجرا می‌شود و بعد تمام می‌شود
    finally:
        engine.stop()         # آزادسازی کامل
        del engine
```

```python
speak_offline("Hello my friend. Welcome to the world of speech processing.")
print("✅ گفتار پخش شد")
```

```python
# نمایش صداهای موجود
engine_temp = pyttsx3.init()
voices = engine_temp.getProperty('voices')
engine_temp.stop()  
del engine_temp

print(f"🔊 تعداد صداهای نصب‌شده روی سیستم: {len(voices)}")
for i, v in enumerate(voices):
    print(f"   {i}: {v.name}")

# انتخاب صدای دوم (اگر وجود داشت) و تنظیم سرعت/بلندی
voice_idx = 1 if len(voices) > 1 else 0
speak_offline(
    text="This is a customized voice with lower speed and higher volume.",
    rate=100,
    volume=0.9,
    voice_index=voice_idx
)
print("✅ گفتار شخصی‌سازی‌شده پخش شد")
```

### 4️⃣ ذخیرهٔ گفتار در فایل

```python
def save_offline(text, filename, rate=200, volume=1.0, voice_index=None):
    engine = pyttsx3.init()
    try:
        engine.setProperty('rate', rate)
        engine.setProperty('volume', volume)
        if voice_index is not None:
            voices = engine.getProperty('voices')
            if 0 <= voice_index < len(voices):
                engine.setProperty('voice', voices[voice_index].id)
        engine.save_to_file(text, filename)
        engine.runAndWait()   
    finally:
        engine.stop()
        del engine
```

### 5️⃣ مقایسه خروجی آنلاین و آفلاین (در صورت دسترسی به اینترنت بین الملل)

```python
sentence = "Python makes text to speech very easy."

save_offline(sentence, "compare_offline.wav")

if gtts_ready:
    try:
        tts = gTTS(sentence, lang='en')
        tts.save("compare_online.mp3")
        print("🔊 مقایسه:")
        print("▶️ آنلاین (Google TTS):")
        display(Audio("compare_online.mp3", autoplay=False))
        print("▶️ آفلاین (pyttsx3):")
        display(Audio("compare_offline.wav", autoplay=False))
    except:
        print("⚠️ اتصال gTTS برقرار نشد. فقط خروجی آفلاین نمایش داده می‌شود:")
        display(Audio("compare_offline.wav", autoplay=False))
else:
    print("🔊 خروجی آفلاین:")
    display(Audio("compare_offline.wav", autoplay=False))
```
