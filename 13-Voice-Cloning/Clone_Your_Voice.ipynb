### 2️⃣ Educational Dashboard 1: Grapheme-to-Phoneme & Acoustic Mapping


```python
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
from g2p_en import G2p

g2p_engine = G2p()

def generate_real_phonemes(text):
    phonemes_list = g2p_engine(text)
    cleaned_phonemes = [p for p in phonemes_list if p.isalnum()]
    return cleaned_phonemes

def synthesize_mock_acoustic_features(phonemes):
    n_mels = 80
    frames_per_phoneme = 15
    spectrogram_parts = []

    for p in phonemes:
        frame = np.random.rand(n_mels, frames_per_phoneme) * 0.1

        if any(char.isdigit() for char in p):
            frame[5:15, :] += 0.8
            frame[25:35, :] += 0.6
            if '1' in p:
                frame *= 1.2
        else:
            if p.startswith(('S', 'Z', 'F', 'V', 'SH', 'TH')):
                frame[50:75, :] += 0.7
            elif p.startswith(('P', 'T', 'K', 'B', 'D', 'G')):
                frame[:, 2:5] += 0.8
            else:
                frame[10:40, :] += 0.3

        spectrogram_parts.append(frame)

    if not spectrogram_parts:
        return np.zeros((n_mels, 10))

    full_spectrogram = np.concatenate(spectrogram_parts, axis=1)

    from scipy.ndimage import gaussian_filter
    full_spectrogram = gaussian_filter(full_spectrogram, sigma=1)

    return full_spectrogram

text_input = widgets.Text(
    value="Hi my name is Mostafa",
    description='📝 Text:',
    layout=widgets.Layout(width='600px')
)
output_area = widgets.Output()

def run_simulator(text):
    with output_area:
        clear_output(wait=True)
        if not text:
            return

        phoneme_list = generate_real_phonemes(text)
        phonemes_str = " ".join(phoneme_list)

        simulated_spec = synthesize_mock_acoustic_features(phoneme_list)

        simulated_spec_db = librosa.amplitude_to_db(simulated_spec, ref=np.max)

        fig, ax = plt.subplots(figsize=(14, 3))
        img = librosa.display.specshow(simulated_spec_db, y_axis='mel', x_axis='time', ax=ax, cmap='magma')
        ax.set_title("Acoustic Model Output (Phoneme-to-Mel-Spectrogram Mapping)")
        plt.colorbar(img, ax=ax, format="%+2.0f dB")
        plt.tight_layout()

        html_output = f"""
        <div style="background: rgba(30, 30, 40, 0.9); padding: 20px; border-radius: 10px; border-left: 4px solid #00d4ff; font-family: 'Courier New', monospace; margin-bottom: 10px;">
            <h4 style="color: #00d4ff; margin-top: 0;">Step 1: Text Input</h4>
            <div style="color: white; font-size: 16px;">"{text}"</div>
            <hr style="border-color: #444;">
            <h4 style="color: #ff0055; margin-top: 0;">Step 2: Neural Grapheme-to-Phoneme (CMU Dict)</h4>
            <div style="color: #ff99bb; font-size: 16px; letter-spacing: 2px; line-height: 1.5;">[{phonemes_str}]</div>
            <hr style="border-color: #444;">
            <h4 style="color: #00ffcc; margin-top: 0;">Step 3: Acoustic Synthesis</h4>
            <div style="color: #ccc; font-size: 14px;">(See generated Spectrogram below) -> Passed to Vocoder for audio generation.</div>
        </div>
        """
        display(HTML(html_output))
        display(fig)
        plt.close(fig)

def on_text_change(change):
    run_simulator(change['new'])

text_input.continuous_update = False
text_input.observe(on_text_change, names='value')

display(text_input, output_area)
run_simulator(text_input.value)
```

### 3️⃣ Educational Dashboard 2: The Neural Conditioning (Injecting Speaker DNA)


```python
import numpy as np
import matplotlib.pyplot as plt
import librosa.display
import ipywidgets as widgets
from IPython.display import display, clear_output
from scipy.ndimage import gaussian_filter

np.random.seed(42)
n_features = 80
time_steps = 100
base_content = np.random.rand(n_features, time_steps) * 0.1

base_content[10:15, 10:40] += 0.8
base_content[25:30, 10:40] += 0.6
base_content[45:50, 10:40] += 0.4

base_content[20:80, 45:50] += 0.7
base_content[30:80, 80:85] += 0.6

base_content[15:20, 60:90] += 0.7
base_content[35:40, 60:90] += 0.5

base_content = gaussian_filter(base_content, sigma=0.8)

def extract_film_parameters(profile_type):

    gamma = np.ones((n_features, 1)) # Scale (Multiplier)
    beta = np.zeros((n_features, 1)) # Shift (Adder)

    if profile_type == "Deep Voice (Male)":
        gamma[0:25, 0] = 1.8
        beta[5:15, 0] = 0.5
        gamma[40:, 0] = 0.4
    elif profile_type == "Bright Voice (Female)":
        gamma[35:70, 0] = 1.8
        beta[40:60, 0] = 0.4
        gamma[0:20, 0] = 0.5
    elif profile_type == "Robotic / Metallic":
        gamma[::8, 0] = 2.5
        beta[::8, 0] = 0.8

    gamma = gaussian_filter(gamma, sigma=2)
    beta = gaussian_filter(beta, sigma=2)
    return gamma, beta

speaker_dropdown = widgets.Dropdown(
    options=["Deep Voice (Male)", "Bright Voice (Female)", "Robotic / Metallic"],
    value="Deep Voice (Male)",
    description='🧬 Select Speaker:',
    style={'description_width': 'initial'}
)
output_area_2 = widgets.Output()

def run_conditioning_simulator(speaker_profile):
    with output_area_2:
        clear_output(wait=True)

        gamma_scale, beta_shift = extract_film_parameters(speaker_profile)

        conditioned_features = (base_content * gamma_scale) + beta_shift

        base_db = librosa.amplitude_to_db(base_content, ref=np.max)
        conditioned_db = librosa.amplitude_to_db(conditioned_features, ref=np.max)

        fig, axes = plt.subplots(1, 4, figsize=(18, 4), gridspec_kw={'width_ratios': [3, 0.8, 0.8, 3]})

        img1 = librosa.display.specshow(base_db, ax=axes[0], cmap='magma', y_axis='mel')
        axes[0].set_title("1. Base Content Matrix (X)\n[Phonemes/Words]")
        axes[0].set_ylabel("Mel Frequency")

        axes[1].imshow(gamma_scale, aspect='auto', cmap='viridis', origin='lower')
        axes[1].set_title("2. Gamma (γ)\n[Scale Multiplier]")
        axes[1].set_xticks([])
        axes[1].set_yticks([])

        axes[2].imshow(beta_shift, aspect='auto', cmap='plasma', origin='lower')
        axes[2].set_title("3. Beta (β)\n[Energy Shift]")
        axes[2].set_xticks([])
        axes[2].set_yticks([])

        img4 = librosa.display.specshow(conditioned_db, ax=axes[3], cmap='magma', y_axis='mel')
        axes[3].set_title("4. Conditioned Output\n[(X * γ) + β]")

        plt.tight_layout()
        display(fig)
        plt.close(fig)

def on_speaker_change(change):
    run_conditioning_simulator(change['new'])

speaker_dropdown.observe(on_speaker_change, names='value')

display(speaker_dropdown, output_area_2)
run_conditioning_simulator(speaker_dropdown.value)
```

### 4️⃣ Capturing Your Voice inside Google Colab

Since Python cannot directly access your local microphone from a cloud server, we use a Javascript bridge to record audio directly inside your browser and pass it to our Colab environment.

> 🎤 **Action:** Run the cell below, click "Record", and read the prompt clearly. We only need 3 to 5 seconds!

```python
import soundfile as sf

def record_audio_colab(filename="reference_voice_raw.webm", duration=5):
    js_code = f"""
    const sleep  = time => new Promise(resolve => setTimeout(resolve, time))
    const b2text = blob => new Promise(resolve => {{
        const reader = new FileReader()
        reader.onloadend = e => resolve(e.srcElement.result)
        reader.readAsDataURL(blob)
    }})
    var record = async () => {{
        stream = await navigator.mediaDevices.getUserMedia({{audio: true}})
        recorder = new MediaRecorder(stream)
        chunks = []
        recorder.ondataavailable = e => chunks.push(e.data)
        recorder.start()

        const div = document.createElement('div');
        div.innerHTML = '<b style="color:red; font-size:20px;">🔴 Recording in progress... Please speak!</b>';
        document.body.appendChild(div);

        await sleep({duration * 1000})

        recorder.stop()
        await sleep(500)
        stream.getTracks().forEach(track => track.stop())
        div.remove();

        blob = new Blob(chunks)
        text = await b2text(blob)
        return text
    }}
    """
    print(f"🎙️ Get ready to speak for {duration} seconds...")
    display(Javascript(js_code))

    audio_string = eval_js('record()')
    print("✅ Recording captured! Processing...")

    audio_bytes = b64decode(audio_string.split(',')[1])
    with open(filename, 'wb') as f:
        f.write(audio_bytes)

    clean_wav_path = "reference_voice.wav"
    audio_data, sr = librosa.load(filename, sr=22050)
    sf.write(clean_wav_path, audio_data, sr)

    return clean_wav_path

ref_audio_path = record_audio_colab(duration=5)

audio_data, sr = librosa.load(ref_audio_path, sr=None)
display(HTML("<hr><b>Your Reference Voice (The DNA for our Clone):</b>"))
display(Audio(audio_data, rate=sr))
```

### 6️⃣ The Magical Moment: Zero-Shot Synthesis

We now have your voice matrix (`reference_voice.wav`) and our AI Engine (`XTTS`).
Let's pass a completely new sentence to the AI and command it to generate the audio using *your* vocal characteristics.

```python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output, Audio
import os

print("--- 🎙️ AI Voice Synthesis Studio Loading ---")

languages_config = {
    "🇬🇧 English": {"code": "en", "text": "Hello, I am an artificial intelligence, and this is a test of my cloned voice."},
    "🇫🇷 French": {"code": "fr", "text": "Bonjour, je suis une intelligence artificielle, et ceci est un test de ma voix clonée."},
    "🇪🇸 Spanish": {"code": "es", "text": "Hola, soy una inteligencia artificial, y esta es una prueba de mi voz clonada."},
    "🇯🇵 Japanese": {"code": "ja", "text": "こんにちは、私は人工知能です。これは私のクローン音声のテストです。"},
    "🇸🇦 Arabic": {"code": "ar", "text": "مرحباً، أنا ذكاء اصطناعي، وهذا اختبار لصوتي المستنسخ."},
    "🇩🇪 German": {"code": "de", "text": "Hallo, ich bin eine künstliche Intelligenz, und dies ist ein Test meiner geklonten Stimme."}
}

lang_dropdown = widgets.Dropdown(
    options=list(languages_config.keys()),
    value="🇫🇷 French",
    description='🌍 Language:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='300px')
)

text_input = widgets.Textarea(
    value=languages_config["🇫🇷 French"]["text"],
    description='✍️ Text:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='600px', height='80px')
)

generate_btn = widgets.Button(
    description='🎙️ Clone My Voice!',
    button_style='success',
    tooltip='Click to synthesize your voice',
    icon='play',
    layout=widgets.Layout(width='200px', height='40px')
)

output_area = widgets.Output()

def on_language_change(change):
    selected_lang = change['new']
    text_input.value = languages_config[selected_lang]["text"]

lang_dropdown.observe(on_language_change, names='value')

def on_generate_click(b):
    with output_area:
        clear_output(wait=True)
        selected_lang = lang_dropdown.value
        lang_code = languages_config[selected_lang]["code"]
        target_text = text_input.value

        if not os.path.exists("reference_voice.wav"):
            print("❌ Error: 'reference_voice.wav' not found! Please run the recording cell first.")
            return

        html_wait = f"""
        <div style="background-color: #2b2b2b; padding: 15px; border-radius: 8px; border-left: 5px solid #f39c12;">
            <b style="color: #f39c12;">⏳ Processing Pipeline Initiated...</b><br>
            <span style="color: #ddd;">1. Extracting Speaker DNA (Scale & Shift parameters) from reference...</span><br>
            <span style="color: #ddd;">2. Translating text to {selected_lang} Phonemes...</span><br>
            <span style="color: #ddd;">3. Conditioning Neural Matrix and generating Audio...</span>
        </div>
        """
        display(HTML(html_wait))

        output_file = f"cloned_{lang_code}.wav"

        try:
            tts.tts_to_file(
                text=target_text,
                speaker_wav="reference_voice.wav",
                language=lang_code,
                file_path=output_file
            )
            clear_output(wait=True)
            html_success = f"""
            <div style="background-color: #2b2b2b; padding: 15px; border-radius: 8px; border-left: 5px solid #2ecc71;">
                <b style="color: #2ecc71;">✅ Zero-Shot Synthesis Successful!</b><br>
                <span style="color: #ddd;">Language: {selected_lang}</span>
            </div>
            """
            display(HTML(html_success))
            display(Audio(output_file, autoplay=True))
            display(HTML(f'<br><a href="{output_file}" download target="_blank" style="color: #3498db; text-decoration: none; font-weight: bold;">📥 Download Cloned Audio (.wav)</a>'))

        except Exception as e:
            clear_output(wait=True)
            print(f"❌ Error during deep learning synthesis: {e}")

generate_btn.on_click(on_generate_click)

title_widget = widgets.HTML(value="<h3>🎛️ Cross-Lingual Zero-Shot Studio</h3>")

ui_box = widgets.VBox([
    title_widget,
    lang_dropdown,
    text_input,
    generate_btn
])

display(ui_box, output_area)
```
